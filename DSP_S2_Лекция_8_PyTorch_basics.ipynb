```python
import torch
from torchviz import make_dot

a = torch.tensor(10.0, dtype=torch.float64, requires_grad=True)
b = torch.tensor(20.0, dtype=torch.float64, requires_grad=True)
c = a + b
print(c.detach().cpu().numpy())

# Visualize the graph that produced the output tensor 'z'
dot = make_dot(c, params=dict(a=a,b=b))
dot.render("graph", format="png")
```

```python
import numpy as np
import torch
N = 1000
x = np.random.randn( N,1 ).flatten()
y = x*5.0
h = torch.tensor(0.1,dtype=torch.float64, requires_grad=True)
mu = 1e-2
for n in range(N):
  # Данные для n-ой итерации
  X = torch.tensor(x[n])
  Y = torch.tensor(y[n])
  # Forward: Выход адаптируемой модели ^y(n)=hx(n)
  yh = h*X
  # Ошибка e(n)
  e = Y - yh
  # Функция потерь Q(n) = e(n)^2
  Q = e*e
  # Градиент
  Q.backward()
  with torch.no_grad():
    # Шаг адаптации
    h -= mu*h.grad
    # Сбрасываем градиент
    h.grad = None

print(h.detach().cpu().numpy())
```
